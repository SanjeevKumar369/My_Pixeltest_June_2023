```python

# Load your Excel file into a DataFrame
file_path = "/content/112_Animal_Husbandry.xlsx"
df = pd.read_excel(file_path)

# Display the DataFrame (optional)
print(df)

# Save the DataFrame to an ODS file
# Note: You might need to convert the DataFrame to a list of lists for this


# Convert DataFrame to a list of lists
data = [df.columns.tolist()] + df.values.tolist()

# Create a new ODS file
sheet = pe.Sheet(data)
sheet.save_as("Animal_Husbandry.ods")

# Download the ODS file
files.download("Animal_Husbandry.ods")
```
